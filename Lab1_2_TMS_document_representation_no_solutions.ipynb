# 1st LAB: Document Representation

Refers to the process of converting raw text or documents into a format that can be understood and processed by machine learning models or algorithms.

**Since machines can't directly work with raw text, we need to transform it into structured numerical representations that capture the underlying meaning and structure of the document.**

## Create sample data

````python
import pandas as pd
import numpy as np

corpus = [
    'The sky is blue and beautiful.',
    'Love this blue and beautiful sky!',
    'The quick brown fox jumps over the lazy dog.',
    "A king's breakfast has sausages, ham, bacon, eggs, toast and beans",
    'I love green eggs, ham, sausages and bacon!',
    'The brown fox is quick and the blue dog is lazy!',
    'The sky is very blue and the sky is very beautiful today',
    'The dog is lazy but the brown fox is quick!'
]

labels = [
    'weather', 'weather', 'animals', 'food', 'food', 'animals', 'weather',
    'animals'
]

corpus = np.array(corpus)
corpus_df = pd.DataFrame({'Document': corpus, 'Category': labels})
corpus_df = corpus_df[['Document', 'Category']]
corpus_df
````

## Count Vectorizer

The Count Vectorizer is one of the most widely used techniques for converting text documents into a structured numerical format. It is part of the **Bag of Words (BoW) model**, which **represents text data by counting the occurrences of words in a document**.

In simpler terms, the **Count Vectorizer creates a vector where each element corresponds to the frequency of a word in a document**.

*This vector can then be used as input to machine learning models for tasks like text classification, sentiment analysis, and information retrieval.*

````python
from sklearn.feature_extraction.text import CountVectorizer

# Initialize CountVectorizer to convert text documents into a matrix of token counts
# min_df=0.0 and max_df=1.0 ensure that all terms are considered, regardless of their document frequency
cv = CountVectorizer(min_df=0., max_df=1.)

# Fit the vectorizer on the 'Document' column and transform the text data into a document-term matrix
cv_matrix = cv.fit_transform(corpus_df.Document)

# Convert the sparse matrix into a dense NumPy array for easier manipulation
cv_matrix = cv_matrix.toarray()
````

````python
# Get all unique words in the corpus
vocab = cv.get_feature_names_out()

# Display the document-term matrix with words as columns
pd.DataFrame(cv_matrix, columns=vocab)
````

### Binary vectorizer

A Binary Vectorizer is a technique used to represent text data in a binary format, where each feature (typically a word) is either present or absent in a document.

The **binary vectorization** approach is a variation of the Bag of Words (BoW) model, but instead of counting the frequency of words, it **only records whether a word appears or not**.

````python
# Count vs Binary on 2 small documents
docs = ["bacon and eggs", "eggs and eggs"]

count = CountVectorizer()
X_count = count.fit_transform(docs).toarray()

# binary=True: 1 if the word is in the document, 0 if not (the count is lost)
X_binary = CountVectorizer(binary=True).fit_transform(docs).toarray()

vocab_docs = count.get_feature_names_out()
display(pd.DataFrame(X_count, columns=vocab_docs, index=['doc 1', 'doc 2']))
display(pd.DataFrame(X_binary, columns=vocab_docs, index=['doc 1', 'doc 2']))
````

## TF-IDF

**TF-IDF (Term Frequency-Inverse Document Frequency) is a statistical measure used to evaluate the importance of a word within a document relative to a collection of documents (also called a corpus).**

*It is commonly used in text mining, information retrieval, and natural language processing (NLP) to transform text into numerical features, which can then be used for machine learning tasks such as classification, clustering, or recommendation.*




````python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize TfidfVectorizer to convert text into TF-IDF vectors
tv = TfidfVectorizer(min_df=0., max_df=1., use_idf=True)

# Transform documents into TF-IDF matrix
tv_matrix = tv.fit_transform(corpus_df.Document)

# Convert the sparse matrix into a dense array
tv_matrix = tv_matrix.toarray()

# Get unique words and display rounded TF-IDF scores
vocab = tv.get_feature_names_out()
pd.DataFrame(np.round(tv_matrix, 2), columns=vocab)
````

````python
# Calculate TF-IDF manually using term frequency and inverse document frequency
n = len(docs)                      # Total number of documents
tf = X_count                       # Term frequency: word count in each document
df = (X_count > 0).sum(axis=0)     # Document frequency: documents containing each word
idf = np.log10(n / df)             # Inverse document frequency

# Display the manually calculated TF-IDF matrix
display(pd.DataFrame(np.round(tf * idf, 2), columns=vocab_docs, index=['doc 1', 'doc 2']))

# Calculate TF-IDF using sklearn's TfidfVectorizer
# It applies smoothed IDF and normalizes each document vector
X_tfidf = TfidfVectorizer().fit_transform(docs).toarray()

# Display the sklearn TF-IDF matrix
display(pd.DataFrame(np.round(X_tfidf, 2), columns=vocab_docs, index=['doc 1', 'doc 2']))
````

### Why are the two tables different?

`eggs` is in **both** documents, so with the formula of the slides its idf is `log10(2/2) = 0`: in doc 2 `eggs` appears twice, but its weight is **0**. The same happens to `and`.

sklearn changes the formula in two ways:

1. **Smoothed idf:** `idf = ln((1 + n) / (1 + df)) + 1`. The `+ 1` at the end means that a word in all the documents gets idf = 1, not 0: it counts less, but it is not removed.
   - `and`, `eggs`: `ln(3/3) + 1 = 1`
   - `bacon`: `ln(3/2) + 1 = 1.41`
2. **Normalization:** each row is divided by its length, so every document vector has length 1. Doc 1 = `(1, 1.41, 1) / 1.99` = `(0.50, 0.70, 0.50)`.

To get the formula without smoothing: `TfidfVectorizer(smooth_idf=False)`; without normalization: `norm=None`.

## Document Similarity

It refers to the task of **measuring how similar two or more text documents are to each other based on their content**. Document similarity is crucial in many NLP applications such as:

- Search Engines: Ranking documents based on relevance to a search query.

- Recommendation Systems: Suggesting similar documents or items.

- Plagiarism Detection: Identifying copied content.

- Text Clustering: Grouping similar documents together.

- Question Answering: Matching user queries to relevant documents.

The basic idea is to convert text documents into a mathematical representation and then apply a similarity measure to quantify how similar they are.

**Cosine similarity is one of the most popular methods for measuring the similarity between two documents. It measures the cosine of the angle between two vectors in a multi-dimensional space. If the vectors are similar (i.e., the angle between them is small), the cosine similarity is close to 1; if they are completely different, the cosine similarity is 0.**

````python
from sklearn.metrics.pairwise import cosine_similarity

# Calculate cosine similarity between all document vectors
similarity_matrix = cosine_similarity(tv_matrix)

# Create labels combining document index and category
labels_df = [f"{i} {c}" for i, c in enumerate(corpus_df.Category)]

# Display the similarity matrix with document labels
similarity_df = pd.DataFrame(
    np.round(similarity_matrix, 2),
    index=labels_df,
    columns=labels_df
)
similarity_df
````

````python
# Compare cosine similarity with and without English stop words
sim_all = cosine_similarity(TfidfVectorizer().fit_transform(corpus_df.Document))
sim_no_stop = cosine_similarity(TfidfVectorizer(stop_words='english').fit_transform(corpus_df.Document))

# Select document pairs to compare
pairs = [(0, 5), (3, 4), (0, 6)]

# Display similarity scores before and after removing stop words
pd.DataFrame({
    'pair': [f"{labels_df[i]} - {labels_df[j]}" for i, j in pairs],
    'with stop words': [sim_all[i, j].round(2) for i, j in pairs],
    'without stop words': [sim_no_stop[i, j].round(2) for i, j in pairs],
})
````

Documents 0 (weather) and 5 (animals) looked similar only because they share `the`, `is`, `and`: *this is why in the first notebook we remove the stop words*.

### Document similarity to a query

In many applications, we are not just interested in comparing documents to each other but also in comparing a document (or a set of documents) to a query.

A query is typically a user's input, which could be a search term, a sentence, or a question, and the goal is to **find the document or documents that are most relevant to that query**.

#### Difference between `fit()`, `transform()` and `fit_transform()`

These methods are used to prepare data before applying Machine Learning algorithms.

- **`fit()`**: learns from the data (e.g. vocabulary and IDF weights in TF-IDF).
- **`transform()`**: converts data into numerical representations using what was previously learned.
- **`fit_transform()`**: combines both operations: learns from the data and transforms it.

#### Example with TF-IDF

```python
# Learn the vocabulary and IDF weights, then transform all documents
docs_tfidf = vectorizer.fit_transform(corpus_df.Document)

# Transform a new query using the same vocabulary and IDF weights
query_tfidf = vectorizer.transform([query])
```

#### Why do we use them differently?

- We use `fit_transform()` on the original documents because the vectorizer needs to learn the vocabulary and calculate IDF weights.
- We use `transform()` on a new query because the vectorizer has already learned these values from the original documents.

**Important:** Using `fit_transform()` on the query would create a new vocabulary and new IDF weights, making the query vectors incompatible with the original document vectors.

**Remember:**
- `fit()` = learn
- `transform()` = convert
- `fit_transform()` = learn + convert

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Initialize TF-IDF vectorizer and transform all documents
vectorizer = TfidfVectorizer()
docs_tfidf = vectorizer.fit_transform(corpus_df.Document)

# Calculate cosine similarity between a query and all documents
def get_tf_idf_query_similarity(vectorizer, docs_tfidf, query):
    # Transform the query into a TF-IDF vector
    query_tfidf = vectorizer.transform([query])

    # Calculate similarity scores and return them as a 1D array
    similarity = cosine_similarity(query_tfidf, docs_tfidf).flatten()
    return similarity

# Rank documents by similarity score (highest first)
def show_ranking(scores, ascending=False):
    ranking = corpus_df[['Document']].assign(score=np.round(scores, 2))
    return ranking.sort_values('score', ascending=ascending)

# Test different queries and display the ranked documents
for query in ['eggs', 'king eggs', 'green eggs']:
    print('Query:', query)
    display(show_ranking(get_tf_idf_query_similarity(vectorizer, docs_tfidf, query)))
````
