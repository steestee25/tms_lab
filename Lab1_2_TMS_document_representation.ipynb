# Document Representation

Refers to the process of converting raw text or documents into a format that can be understood and processed by machine learning models or algorithms. Since machines can't directly work with raw text, we need to transform it into structured numerical representations that capture the underlying meaning and structure of the document.

### Create sample data

```python
import pandas as pd
import numpy as np

corpus = [
    'The sky is blue and beautiful.',
    'Love this blue and beautiful sky!',
    'The quick brown fox jumps over the lazy dog.',
    "A king's breakfast has sausages, ham, bacon, eggs, toast and beans",
    'I love green eggs, ham, sausages and bacon!',
    'The brown fox is quick and the blue dog is lazy!',
    'The sky is very blue and the sky is very beautiful today',
    'The dog is lazy but the brown fox is quick!'
]

labels = [
    'weather', 'weather', 'animals', 'food', 'food', 'animals', 'weather',
    'animals'
]

corpus = np.array(corpus)
corpus_df = pd.DataFrame({'Document': corpus, 'Category': labels})
corpus_df = corpus_df[['Document', 'Category']]
corpus_df
```

### Binary vectorizer

A Binary Vectorizer is a technique used to represent text data in a binary format, where each feature (typically a word) is either present or absent in a document. The binary vectorization approach is a variation of the Bag of Words (BoW) model, but instead of counting the frequency of words, it only records whether a word appears or not.

### Count Vectorizer

The Count Vectorizer is one of the most widely used techniques for converting text documents into a structured numerical format. It is part of the Bag of Words (BoW) model, which represents text data by counting the occurrences of words in a document. In simpler terms, the Count Vectorizer creates a vector where each element corresponds to the frequency of a word in a document. This vector can then be used as input to machine learning models for tasks like text classification, sentiment analysis, and information retrieval.

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(min_df=0., max_df=1.)
cv_matrix = cv.fit_transform(corpus_df.Document.iloc[:100]) #top 100 documents (not required here)
cv_matrix = cv_matrix.toarray()
print(cv_matrix)
```

```python
# get all unique words in the corpus
vocab = cv.get_feature_names_out()

# show document feature vectors
pd.DataFrame(cv_matrix, columns=vocab)
```

###TF_IDF

TF-IDF (Term Frequency-Inverse Document Frequency) is a statistical measure used to evaluate the importance of a word within a document relative to a collection of documents (also called a corpus). It is commonly used in text mining, information retrieval, and natural language processing (NLP) to transform text into numerical features, which can then be used for machine learning tasks such as classification, clustering, or recommendation.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Setting the vectorizer
tv = TfidfVectorizer(min_df=0., max_df=1., use_idf=True)

# Transforming the data to vectors
tv_matrix = tv.fit_transform(corpus_df.Document)
tv_matrix = tv_matrix.toarray()

vocab = tv.get_feature_names_out()
pd.DataFrame(np.round(tv_matrix, 2), columns=vocab)
```

### Document Similarity

It refers to the task of measuring how similar two or more text documents are to each other based on their content. Document similarity is crucial in many NLP applications such as:

- Search Engines: Ranking documents based on relevance to a search query.

- Recommendation Systems: Suggesting similar documents or items.

- Plagiarism Detection: Identifying copied content.

- Text Clustering: Grouping similar documents together.

- Question Answering: Matching user queries to relevant documents.

The basic idea is to convert text documents into a mathematical representation and then apply a similarity measure to quantify how similar they are.

Cosine similarity is one of the most popular methods for measuring the similarity between two documents. It measures the cosine of the angle between two vectors in a multi-dimensional space. If the vectors are similar (i.e., the angle between them is small), the cosine similarity is close to 1; if they are completely different, the cosine similarity is 0.

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(tv_matrix)
similarity_df = pd.DataFrame(similarity_matrix)
similarity_df
```

# Document similarity to a query

In many applications, we are not just interested in comparing documents to each other but also in comparing a document (or a set of documents) to a query. A query is typically a user's input, which could be a search term, a sentence, or a question, and the goal is to find the document or documents that are most relevant to that query.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer()
docs_tfidf = vectorizer.fit_transform(corpus_df.Document)

def get_tf_idf_query_similarity(vectorizer, docs_tfidf, query):
    """
    vectorizer: TfIdfVectorizer model
    docs_tfidf: tfidf vectors for all docs
    query: query doc
    return: cosine similarity between query and all docs
    """
    query_tfidf = vectorizer.transform([query])

    similarity = cosine_similarity(query_tfidf, docs_tfidf).flatten()
    return similarity


display(get_tf_idf_query_similarity(vectorizer, docs_tfidf, 'eggs'))
display(get_tf_idf_query_similarity(vectorizer, docs_tfidf, 'king eggs'))
display(get_tf_idf_query_similarity(vectorizer, docs_tfidf, 'green eggs'))

corpus_df.head(10)
```
